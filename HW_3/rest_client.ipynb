```
import requests
import pandas as pd
```

```
host = "https://b4ef5aa4-3f71-4837-91f1-c6fd9810a7ea.deepnoteproject.com"
mac_address = "0xe45f01e89914"
start_date = "2025-01-20"
end_date = "2025-01-26"
```

```
# Check the service status
response = requests.get(host + '/status')

if response.status_code == 200:
    status = response.json()['status']
    print(f'The server is {status}.')
else:
    print('The server is offline.')
```

```
# Add RPI sensor node
payload = {'mac_address': mac_address}
response = requests.post(host + '/sensors', json=payload)

if response.status_code == 200:
    print()
    print('Sensor timeseries added.')
else:
    print(response.status_code, response.reason)
```

```
# Retrieve data from sensor in a given time range
payload = {"start_date": start_date,
           "end_date": end_date}
response = requests.get(host + f'/data/{mac_address}', params=payload)

if response.status_code == 200:
    print()
    print('Everything worked as expected.')

    data = response.json()
    mac_address = data['mac_address']
    timestamp_list = data['timestamp']
    temperature_list = data['temperature']
    humidity_list = data['humidity']
else:
    print(response.status_code, response.reason)
```

```
df_cols = [*zip(timestamp_list, temperature_list, humidity_list)]
df = pd.DataFrame(df_cols, columns=['Timestamp', 'Temperature', 'Humidity'])
```
